```python
import pyarrow as pa
import pyarrow.parquet as par

import matplotlib.pyplot as plt
```

```python
pf = par.ParquetFile("src/ticks/date=2025-12-03/part-00000.parquet")
```

```python
ticks = pf.read()
```

```python
ticks_df = ticks.to_pandas()
```

```python
ticks_df.name.unique()
```

```python
%store -r instruments
```
