```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import kiteconnect
from datetime import datetime
from ticks_collector.kite_utils import get_initialized_kite
from utils import get_historical_data
```

# Download Data

```python
kite = get_initialized_kite()
```

```python
nse_ins = kite.instruments(exchange="NSE")
```

```python
nse_ins = {i['tradingsymbol']: i for i in nse_ins}
```

```python
nifty_total_market = pd.read_csv("data/ind_niftytotalmarket_list.csv")
```

```python
nifty_total_market['instrument_token'] = nifty_total_market.Symbol.apply(lambda x: nse_ins[x]['instrument_token'] if x in nse_ins else pd.NA)
```

```python
nifty_total_market.dropna(subset=["instrument_token"], inplace=True)
```

```python
end = datetime.today()
start = datetime(year=2025, month=4, day=1)
interval = "day"

for _, ins in nifty_total_market.iterrows():
    if pd.isna(ins.instrument_token):
        print(f"NA ins token for {ins.Symbol}, skipping...")
        continue
    data = get_historical_data(kite, ins.instrument_token, interval, start, end)
    data = pd.DataFrame(data)
    data.to_csv(f"data/historical_data_{start.strftime('%y-%m-%d')}_{end.strftime('%y-%m-%d')}_day/{ins.Symbol}_{start.strftime('%y-%m-%d')}_{end.strftime('%y-%m-%d')}_day.csv", index=False, mode='x')
```

# Loading data

```python
nifty_total_market['data_path'] = nifty_total_market.apply(lambda x: f"data/historical_data_{start.strftime('%y-%m-%d')}_{end.strftime('%y-%m-%d')}_day/{x.Symbol}_{start.strftime('%y-%m-%d')}_{end.strftime('%y-%m-%d')}_day.csv" if not pd.isna(x.instrument_token) else pd.NA, axis=1)
```
