```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
def random_walk(start_point, count, std, seed=1):
    # np.random.seed(seed)
    y = np.zeros(count)
    y[0] = start_point
    for i in range(1, count):
        y[i] = y[i-1]*(1.0001) + (np.random.rand() - 0.5)*std

    return y
    
```

```python
fig = plt.figure(figsize=(20, 10))
ax = fig.subplots()
ax.plot(random_walk(161, 10000, 10))
```
